## Motivation — Quantifying the Optimal Conditions for Aurora Sightings

The comparison between high-latitude auroral observatories and tourist destinations naturally leads to a central question:

> **How can we objectively quantify the best *time–location* combinations to observe the Northern Lights?**

While iconic locations such as Yellowknife, Kiruna, or Abisko are well known, their relative performance is often discussed qualitatively (latitude, reputation, climate) rather than through a reproducible, data-driven framework.

---

## Initial Approach: Keogram-Based Analysis (High Fidelity, High Cost)

One possible approach is to rely directly on **all-sky camera keograms**, which provide a physically grounded, time-resolved record of auroral activity.

For example, keograms are available for:

* Kiruna (Sweden)
* Yellowknife (Canada)

However, a keogram-centric pipeline rapidly becomes a **heavy processing project**, requiring:

* Bulk download of image archives (multi-year)
* Image normalization and filtering
* Separation of night/day segments
* Cloud obstruction handling
* Temporal alignment with geomagnetic indices (Kp)
* Aggregation over lunar phase and local illumination conditions

While scientifically robust, this approach is **time-consuming and location-dependent**, making it poorly suited as a first-order comparative tool across many sites.

---

## A More Scalable Baseline: Aggregated Environmental & Geophysical Data

Before introducing image-level validation, a **lighter and more general approach** can already answer much of the question.

The following variables are:

* well documented,
* available globally,
* and directly relevant to auroral visibility.

### Core Variables

| Factor                        | Role in Visibility                          |
| ----------------------------- | ------------------------------------------- |
| **Nighttime duration**        | Defines the available observation window    |
| **Cloud coverage**            | Primary visibility limiter                  |
| **Geomagnetic activity (Kp)** | Proxy for auroral probability and intensity |
| **Moon phase & altitude**     | Affects sky brightness                      |
| **Latitude / city location**  | Determines proximity to the auroral oval    |

These datasets can be **aggregated statistically** over long periods (multiple seasons or years) to estimate:

* Probability of “aurora-favorable” nights
* Seasonal and monthly optimal windows
* Relative performance between locations

This already enables a **comparative ranking of sites**, answering questions such as:

* *Where is it statistically easier to see auroras without chasing?*
* *When is the optimal season for a given location?*

---

## Progressive Refinement: Keogram Validation and Scoring

Once this baseline analysis is established, keograms can be reintroduced as a **second-stage validation tool**:

* Verify whether aggregated predictors correlate with actual auroral presence
* Quantify false positives (good conditions, no aurora)
* Derive a **visibility score** combining:

  * geomagnetic forcing,
  * sky conditions,
  * and observed auroral occurrence

This opens the door to:

* Cross-site calibration (e.g. Yellowknife vs Kiruna)
* Objective comparison between North America and Europe
* Eventually, **predictive scoring of optimal travel windows**

---

## Conceptual Outcome

The final objective is not merely to detect auroras, but to:

> **Quantify the probability of *seeing* auroras at a given place and time, under realistic observational constraints.**

By separating:

1. **Low-cost statistical aggregation**, and
2. **High-fidelity keogram confirmation**,

the analysis remains both **scalable** and **scientifically grounded**, while staying directly relevant to real-world aurora sighting decisions.

#### 2.2.1 Night duration
* Sun altitude < **−18°** (astronomical night)



---

````python
from astral import LocationInfo
from astral.sun import sun
from datetime import date, timedelta
import pytz

from astral.sun import sun
from datetime import timedelta
import pytz

def nautical_night_duration_hours(loc, d, depression=12):
    """
    Returns duration (hours) where Sun altitude < -depression deg.
    Robust to polar/twilight edge cases.
    """
    tz = pytz.timezone(loc.timezone)

    # Primary path: use dusk on day d and dawn on day d+1
    try:
        s1 = sun(loc.observer, date=d, tzinfo=tz, dawn_dusk_depression=depression)
        s2 = sun(loc.observer, date=d + timedelta(days=1), tzinfo=tz, dawn_dusk_depression=depression)

        dusk = s1["dusk"]
        dawn = s2["dawn"]

        return (dawn - dusk).total_seconds() / 3600

    except ValueError as e:
        msg = str(e).lower()

        # Case A: Sun never reaches -depression => no nautical night
        if "never reaches" in msg and "below the horizon" in msg:
            return 0.0

        # Otherwise: re-raise (unexpected failure)
        raise


loc = LocationInfo(
    name="Yellowknife",
    region="Canada",
    timezone="America/Yellowknife",
    latitude=62.4540,
    longitude=-114.3718,
)

print(nautical_night_duration_hours(loc, date(2026, 1, 23), depression=18))
````

````python
from datetime import date, datetime, time
from pathlib import Path

import pandas as pd


def get_wind_and_elevation(loc, d, hour, data_dir=None):
    """
    Return wind speed (km/h) and elevation (m) for a location at date + hour.

    loc: "kiruna" or "yellowknife"
    d: datetime.date
    hour: int (0-23)
    """
    loc_key = loc.strip().lower()
    file_map = {
        "kiruna": "kiruna_cloud_wind_1998_2026.csv",
        "yellowknife": "yellowknife_cloud_wind_1998_2026.csv",
    }
    if loc_key not in file_map:
        raise ValueError(f"Unknown location: {loc}")

    base = Path(data_dir) if data_dir else Path("/media/nuopel/DATA/ProjectsAndTools/1_Projects/AI/Capstone_2/Notebooks/Ext2_Auroravisibility/Data")
    path = base / file_map[loc_key]

    with path.open() as f:
        header = f.readline().strip().split(',')
        values = f.readline().strip().split(',')
    meta = dict(zip(header, values))
    elevation_m = float(meta["elevation"])

    df = pd.read_csv(path, skiprows=3)
    target = datetime.combine(d, time(hour=hour))
    ts = target.strftime("%Y-%m-%dT%H:%M")

    row = df.loc[df["time"] == ts]
    if row.empty:
        raise ValueError(f"No data for {loc_key} at {ts}")

    wind_kmh = float(row["wind_speed_10m (km/h)"].iloc[0])

    return {
        "location": loc_key,
        "datetime": target,
        "wind_speed_10m_kmh": wind_kmh,
        "elevation_m": elevation_m,
    }


# Test: extract wind + elevation for 2026-01-23 at 00:00
print(get_wind_and_elevation("kiruna", date(2026, 1, 20), 0))
print(get_wind_and_elevation("yellowknife", date(2026, 1, 20), 0))
````


#### Geomagnetic activity (Kp)

* Already available from prior work.
* Night aggregation:

  * **max Kp** → favors short intense events
  * **mean Kp** → favors sustained activity
* Choice must be documented and frozen.


````python
from datetime import date, datetime, time
from pathlib import Path

import pandas as pd

PROJECT_ROOT = Path.cwd()
DATA_DIR = PROJECT_ROOT / "../../Data"

KP_FILE = DATA_DIR / "kp_ap.txt"


def load_kp_spine() -> pd.DataFrame:
    columns = [
        "year",
        "month",
        "day",
        "hour_start",
        "hour_mid",
        "days_start",
        "days_mid",
        "kp",
        "ap",
        "definitive_flag",
    ]
    df = pd.read_csv(KP_FILE, sep=r"\s+", header=None, names=columns, comment="#")

    hour_float = df["hour_start"]
    hours = hour_float.astype(int)
    minutes = ((hour_float - hours) * 60).round().astype(int)

    df["timestamp_utc"] = pd.to_datetime(
        {
            "year": df["year"],
            "month": df["month"],
            "day": df["day"],
            "hour": hours,
            "minute": minutes,
        },
        utc=True,
    )

    kp_df = df[["timestamp_utc", "kp"]].copy()
    kp_df = kp_df.sort_values("timestamp_utc").reset_index(drop=True)
    return kp_df


def get_kp_at(d, hour, kp_df):
    """
    Return Kp value for a date + hour (UTC).

    d: datetime.date
    hour: int (0-23)
    """
    target = datetime.combine(d, time(hour=hour))
    ts = pd.Timestamp(target, tz="UTC")

    row = kp_df.loc[kp_df["timestamp_utc"] == ts]
    if row.empty:
        raise ValueError(f"No Kp data for {ts}")

    return float(row["kp"].iloc[0])


kp_df = load_kp_spine()

# Test: extract Kp for 2026-01-23 at 00:00 UTC
print(get_kp_at(date(2025, 1, 23), 0, kp_df))

````

#### Moon parameters

* Computed locally (no API dependency).
* Libraries:

  * **Skyfield** or **PyEphem**
* Hourly sampling within night window:

  * Moon illumination ∈ [0, 1]
  * Moon altitude (degrees)
* Aggregate to:

  * mean or max altitude,
  * mean illumination,
  * optionally fraction of night with Moon above horizon.

Used to model **sky brightness penalty**.

Reference cross-check:

* timeanddate.com (sanity validation only)

---




````python
from datetime import datetime, date, timedelta, time, timezone
import ephem

LOCATIONS = {
"kiruna": dict(lat="67.8500", lon="20.2500", elev=546),
"yellowknife": dict(lat="62.4540", lon="-114.3718", elev=206),
  "reykjavik": dict(lat="64.1466", lon="-21.9426", elev=0),
"fortsmith": dict(lat="60.00", lon="-111.87", elev=205),
}
def moon_series(loc_key, d, tz=timezone.utc):
    loc = LOCATIONS[loc_key]
    obs = ephem.Observer()
    obs.lat = loc["lat"]
    obs.lon = loc["lon"]
    obs.elev = loc["elev"]

    # Build 24 hourly samples for that date
    rows = []
    for h in range(24):
      dt = datetime.combine(d, time(h), tzinfo=tz)
      obs.date = dt  # UTC
      moon = ephem.Moon(obs)
      alt_deg = float(moon.alt) * 180.0 / ephem.pi
      illum = float(moon.moon_phase)  # 0..1
      rows.append((dt, alt_deg, illum))
    return rows

def f_moon_above_h0(loc_key, d, h0_deg=0.0, tz=timezone.utc):
    rows = moon_series(loc_key, d, tz=tz)
    above = sum(1 for _, alt, _ in rows if alt > h0_deg)
    return above / len(rows)

# Example usage
rows_kiruna = moon_series("kiruna", date(2026, 1, 23))
rows_yk = moon_series("yellowknife", date(2026, 1, 23))

print(rows_kiruna[0])  # (timestamp, altitude_deg, illumination)
print(rows_yk[0])
````

````python
import pandas as pd
from astral import LocationInfo
from astral.sun import elevation

LOC_INFO = {
  "kiruna": LocationInfo(
      name="Kiruna",
      region="Sweden",
      timezone="Europe/Stockholm",
      latitude=67.8500,
      longitude=20.2500,
  ),
  "yellowknife": LocationInfo(
      name="Yellowknife",
      region="Canada",
      timezone="America/Yellowknife",
      latitude=62.4540,
      longitude=-114.3718,
  ),
    "reykjavik": LocationInfo(
        name="Reykjavik",
        region="Iceland",
        timezone="Atlantic/Reykjavik",
        latitude=64.1466,
        longitude=-21.9426,
    ),
    "fortsmith": LocationInfo(
      name="Fort Smith",
      region="Canada",
      timezone="America/Yellowknife",
      latitude=60.0000,
      longitude=-111.8833,
  ),
}

def is_astronomical_night(ts_utc, loc, depression=18):
  # Astral expects aware datetime; use UTC directly
  alt_deg = elevation(loc.observer, ts_utc)
  return alt_deg < -depression

def add_is_night(df, location_key):
  loc = LOC_INFO[location_key]
  return df.assign(
      location=location_key,
      is_night=df["timestamp_utc"].apply(lambda ts: is_astronomical_night(ts, loc))
  )

kiruna = add_is_night(date_df.copy(), "kiruna")
yellowknife = add_is_night(date_df.copy(), "yellowknife")
reykjavik = add_is_night(date_df.copy(), "reykjavik")
fortsmith = add_is_night(date_df.copy(), "fortsmith")
````

````python
from datetime import datetime, time, timezone
import ephem
import pandas as pd

LOCATIONS = {
  "kiruna": dict(lat="67.8500", lon="20.2500", elev=546),
  "yellowknife": dict(lat="62.4540", lon="-114.3718", elev=206),
  "reykjavik": dict(lat="64.1466", lon="-21.9426", elev=0),
"fortsmith": dict(lat="60.00", lon="-111.87", elev=205),
}

def get_moon_alt_illum(loc_key, ts_utc):
  loc = LOCATIONS[loc_key]
  obs = ephem.Observer()
  obs.lat = loc["lat"]
  obs.lon = loc["lon"]
  obs.elev = loc["elev"]

  obs.date = ts_utc  # UTC
  moon = ephem.Moon(obs)
  alt_deg = float(moon.alt) * 180.0 / ephem.pi
  illum = float(moon.moon_phase)
  return alt_deg, illum

def add_moon_night(df, loc_key):
  df = df.copy()
  # compute only for night rows
  def compute(row):
      if not row["is_night"]:
          return pd.Series({"moon_altitude_deg": pd.NA, "moon_illumination": pd.NA})
      alt, illum = get_moon_alt_illum(loc_key, row["timestamp_utc"].to_pydatetime())
      return pd.Series({"moon_altitude_deg": alt, "moon_illumination": illum})

  moon_cols = df.apply(compute, axis=1)
  return pd.concat([df, moon_cols], axis=1)

kiruna = add_moon_night(kiruna, "kiruna")
yellowknife = add_moon_night(yellowknife, "yellowknife")
reykjavik = add_moon_night(reykjavik, "reykjavik")
fortsmith = add_moon_night(fortsmith, "fortsmith")

````

````python
from pathlib import Path
import numpy as np
import pandas as pd

````

````python
# Constants (tune later if needed)
H0 = 0.0   # moon altitude threshold (deg)
I0 = 0.25  # moon illumination threshold
K0 = 2.5   # start of meaningful aurora chance
K1 = 5.0   # strong activity
BETA = 1.5 # cloud softness

def cloud_factor(cloud_cover):
  base = (60.0 - cloud_cover) / 30.0
  base = np.clip(base, 0.0, 1.0)
  return base ** BETA

def moon_factor(df):
    # Moon present indicator
    m_up = (df["moon_altitude_deg"] > H0).astype(float)

    # Effective illumination
    illum = df["moon_illumination"]
    i_eff = (illum - I0) / (1.0 - I0)
    i_eff = i_eff.clip(lower=0.0, upper=1.0)

    f_moon = 1.0 - (m_up * i_eff)

    # Missing moon info at night -> neutral (1.0) + QA flag
    missing_night = df["is_night"] & (df["moon_altitude_deg"].isna() | df["moon_illumination"].isna())
    f_moon = f_moon.where(~missing_night, 1.0)
    return f_moon, missing_night

def kp_factor(kp):
  return np.clip((kp - K0) / (K1 - K0), 0.0, 1.0)

def score_hourly(df):
    df = df.copy()
    df["timestamp_utc"] = pd.to_datetime(df["timestamp_utc"], utc=True)

    df["D"] = df["is_night"].astype(float)
    df["F_cloud"] = cloud_factor(df["cloud_cover"])

    f_moon, missing_night = moon_factor(df)
    df["F_moon"] = f_moon
    # df["moon_missing_night"] = missing_night

    df["F_kp"] = kp_factor(df["kp"])

    df["visibility_score"] = df["D"] * df["F_cloud"] * df["F_moon"] * df["F_kp"]
    return df
````
