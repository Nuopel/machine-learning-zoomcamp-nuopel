```python

import pandas as pd
import numpy as np
import os
import sys
from pathlib import Path
```

## LOT 1.2 Data Loading and first observations

```python
data_path = '../data/france_wines_type1.csv'

if not os.path.exists(data_path):
    print(f"❌ Error: Data file not found at {data_path}")
    sys.exit(1)

df = pd.read_csv(data_path)

print(f"  ✅ Loaded: {data_path}")
print(f"  📈 Shape: {df.shape[0]:,} rows × {df.shape[1]} columns\n")
df
```

```python
df.columns = df.columns.str.lower().str.replace(' ', '_')
df.info()
```

```python
df.describe()
```

```python
# Sort columns: numeric first, then non-numeric
numeric_cols = df.select_dtypes(include=['number']).columns
categorical_cols = df.select_dtypes(exclude=['number']).columns
ordered_cols = list(numeric_cols) + list(categorical_cols)

# Explore unique values, missing values, and data types for each column
for col in ordered_cols:
    dtype = df[col].dtype
    missing = df[col].isna().sum()
    unique = df[col].nunique()
    sample = df[col].dropna().unique()[:5]

    print(f"{col} ({dtype})")
    print(f"  Unique: {unique:,} | Missing: {missing:,}")
    print(f"  Sample: {list(sample)}")
    print()
```

```python
# Detect non-numeric entries in vintage_year
invalid_mask = ~df['vintage_year'].astype(str).str.fullmatch(r'\d{4}')
invalid_values = df.loc[invalid_mask, 'vintage_year']

print("🧩 Non-numeric / invalid 'vintage_year' entries:")
print(invalid_values.value_counts())
# print(f"\nTotal invalid: {invalid_values.count()} / {len(df)}")
```

In this dataset we have both numerical and categorical :
### **Numerical Columns**
* **id**: Unique wine identifier across all vintages
* **id_vintage**: Unique identifier for specific wine-vintage combination
* **type_id**: Wine type category (1 = red wine)

* **global_rating**: Average rating across all vintages (scale 1-5)
* **global_rating_count**: Total number of user ratings for the wine
* **vintage_rating**: Average rating for specific vintage (scale 1-5)
* **vintage_rating_count**: Number of user ratings for specific vintage

* **price_availability**: Wine price in euros (€)

* **structure_acidity**: Acidity level (scale 1-5, where 5 = most acidic) (soft-acid)
* **structure_fizziness**: Carbonation level (scale 1-5, all null for red wines)
* **structure_intensity**: Flavor intensity (scale 1-5, where 5 = most intense) (light-intense)
* **structure_sweetness**: Sweetness level (scale 1-5, where 5 = sweetest) (dry-sweet)
* **structure_tannin**: Tannin level (scale 1-5, where 5 = highest tannin)

## **Categorical Columns**
* **wine**: Commercial name of the wine
* **winery**: Producer/winery name
* **url**: Vivino product page URL
* **region**: French wine-producing region (e.g., languedoc-roussillon, saint-estephe)
* **country**: Country of origin (france)
* **vintage_year**: Year of harvest/production

* **grapes**: List of grape varieties with IDs in format [(id, 'name'), ...]
* **food_pairings**: Recommended food pairings with IDs in format [(id, 'name'), ...]

### Handling missing values
There are few values missing, so let's just remove those out of the dataset.

**Negligible Missing Data (can safely ignore):**
- **winery**: 20 missing (0.02%)
- **vintage_year**: 2 missing (0.002%)
- **structure_acidity, structure_intensity, structure_sweetness, structure_tannin**: 4 missing each (0.004%)

**Moderate Missing Data:**
- **grapes**: 4,322 missing (4.7%) - grape composition not documented for some wines

**Complete Missing Data:**
- **structure_fizziness**: 91,060 missing (100%) - expected, as red wines aren't sparkling; **can be dropped**

**Summary:**
Apart from `structure_fizziness` (which should be dropped), all other missing values are minimal (<5%). The dataset is quite complete overall.
I'll replace grape missing values with "Unknown" for now.

I think few columns are not particularly interesting like:
  - food_pairings
  - url
  - country: all from France
  - id and id_vintage : id from vivino, does not make sens to keep them here
  - global_rating and global_rating_count, i'm interest in each wine bottle per year
So i'll drop them
Few categories are identifiers like id, id_vintage, wine and winery.
 vintage_year is an object instead of a int or float



```python
# Drop structure_fizziness column (100% missing)
df_france = df.drop(columns=['structure_fizziness'])
df_france = df_france.drop(columns=['food_pairings'])
df_france = df_france.drop(columns=['url'])
df_france = df_france.drop(columns=['country'])
df_france = df_france.drop(columns=['id'])
df_france = df_france.drop(columns=['type_id'])
df_france = df_france.drop(columns=['id_vintage'])
df_france = df_france.drop(columns=['global_rating'])
df_france = df_france.drop(columns=['global_rating_count'])
df['vintage_year'] = df['vintage_year'].replace('N.V.', np.nan) # replace year missing with nan
print(f"✅ Dropped 'structure_fizziness','food_pairings','url', 'country','country','id', 'id_vintage','global_rating_count','global_rating'")

# Remove rows with any missing values
df_france['grapes'] = df_france['grapes'].fillna('Unknown')
df_france_clean = df_france.dropna()
print(f"✅ Removed rows with missing values")

print(f"📊 Original shape: {df_france.shape[0]:,} rows × {df_france.shape[1]} columns")
print(f"📊 Clean shape: {df_france_clean.shape[0]:,} rows × {df_france_clean.shape[1]} columns")
print(f"🗑️  Removed: {df_france.shape[0] - df_france_clean.shape[0]:,} rows ({((df_france.shape[0] - df_france_clean.shape[0]) / df_france.shape[0] * 100):.1f}%)")
print(f"\n✓ No missing values remaining: {df_france_clean.isna().sum().sum() == 0}")

# Save the clean dataset
output_path = '../data/france_wines_type1_clean.csv'
df_france_clean.to_csv(output_path, index=False)
print(f"\n💾 Saved clean dataset to: {output_path}")
```
